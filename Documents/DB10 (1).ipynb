```python
from torch.utils.data import Dataset
import torchvision.transforms as transforms
import cv2
from PIL import Image

# 🔹 Grayscale (input / LR)
transform_input = transforms.Compose([
    transforms.ToPILImage(),
    transforms.Resize((64, 64)),
    transforms.ToTensor()          # shape: [1, 64, 64]
])

# 🔹 Color (target / HR)
transform_target = transforms.Compose([
    transforms.ToPILImage(),
    transforms.Resize((256, 256)),
    transforms.ToTensor()          # shape: [3, 256, 256]
])

class RemoteSensingDataset(Dataset):
    def __init__(self, image_paths):
        self.image_paths = image_paths

    def __len__(self):
        return len(self.image_paths)

    def __getitem__(self, idx):
        # Read image (color)
        img = cv2.imread(self.image_paths[idx])

        if img is None:
            raise ValueError(f"Could not read image {self.image_paths[idx]}")

        # Convert BGR → RGB
        img_rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)

        # Convert RGB → Grayscale (input)
        img_gray = cv2.cvtColor(img_rgb, cv2.COLOR_RGB2GRAY)

        # Apply transforms
        lr = transform_input(img_gray)     # Black & White input
        hr = transform_target(img_rgb)     # Color target

        return lr, hr
```

```python
from torch.utils.data import Dataset
import torchvision.transforms as transforms
import cv2
import random
import torch

# 🔹 Grayscale input (LR)
transform_lr = transforms.Compose([
    transforms.ToPILImage(),
    transforms.Resize((64, 64)),
    transforms.ToTensor()          # [1, 64, 64]
])

# 🔹 Color target (HR)
transform_hr = transforms.Compose([
    transforms.ToPILImage(),
    transforms.Resize((256, 256)),
    transforms.ToTensor()          # [3, 256, 256]
])

class RemoteSensingDataset(Dataset):
    def __init__(self, image_paths):
        self.image_paths = image_paths

    def __len__(self):
        return len(self.image_paths)

    def __getitem__(self, idx):
        max_attempts = 10

        for _ in range(max_attempts):
            img_path = self.image_paths[idx]
            img = cv2.imread(img_path)

            if img is None:
                # Try another image index
                idx = random.randint(0, len(self.image_paths) - 1)
                continue

            # BGR → RGB
            img_rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)

            # RGB → Grayscale (input)
            img_gray = cv2.cvtColor(img_rgb, cv2.COLOR_RGB2GRAY)

            # Apply transforms
            lr = transform_lr(img_gray)
            hr = transform_hr(img_rgb)

            return lr, hr

        # If all attempts fail → raise error (better than silent failure)
        raise RuntimeError("❌ Failed to load a valid image after multiple attempts.")
```

```python
import torch
import torch.nn as nn
import torch.optim as optim

device = torch.device("cuda" if torch.cuda.is_available() else "cpu")

class SimpleSRColor(nn.Module):
    def __init__(self):
        super().__init__()

        self.encoder = nn.Sequential(
            nn.Conv2d(1, 64, 3, padding=1),   # 1-channel input (grayscale)
            nn.ReLU(),
            nn.Conv2d(64, 128, 3, padding=1),
            nn.ReLU()
        )

        # Upsample from 64x64 → 256x256
        self.upsample = nn.Upsample(scale_factor=4, mode="bilinear", align_corners=False)

        self.decoder = nn.Sequential(
            nn.Conv2d(128, 64, 3, padding=1),
            nn.ReLU(),
            nn.Conv2d(64, 3, 3, padding=1)    # 3-channel output (RGB)
        )

    def forward(self, x):
        x = self.encoder(x)
        x = self.upsample(x)
        x = self.decoder(x)
        return x

model = SimpleSRColor().to(device)

criterion = nn.L1Loss()
optimizer = optim.Adam(model.parameters(), lr=1e-4)
```

```python
import torch.nn as nn

class RSINet(nn.Module):
    def __init__(self):
        super(RSINet, self).__init__()

        # 🔹 Feature extraction (from grayscale)
        self.feature_extractor = nn.Sequential(
            nn.Conv2d(1, 64, kernel_size=9, padding=4),
            nn.ReLU(inplace=True),
            nn.Conv2d(64, 32, kernel_size=5, padding=2),
            nn.ReLU(inplace=True)
        )

        # 🔹 Upsampling (64×64 → 256×256)
        self.upsample = nn.Upsample(scale_factor=4, mode="bilinear", align_corners=False)

        # 🔹 Color reconstruction
        self.reconstructor = nn.Sequential(
            nn.Conv2d(32, 64, kernel_size=3, padding=1),
            nn.ReLU(inplace=True),
            nn.Conv2d(64, 3, kernel_size=3, padding=1)
        )

    def forward(self, x):
        x = self.feature_extractor(x)
        x = self.upsample(x)
        x = self.reconstructor(x)
        return x
```

```python
model = RSINet().to(device)
print("✅ RSINet model initialized on", device)
```

```python
import torch.optim as optim

criterion = nn.L1Loss()  # Best baseline for SR + color
optimizer = optim.Adam(model.parameters(), lr=1e-4)
scheduler = optim.lr_scheduler.CosineAnnealingLR(optimizer, T_max=10)
```
